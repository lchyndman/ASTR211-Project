### About FITS

FITS or FIT files are a common way of storing information for astronomical images. These files have a "header" and "data". The "header" has the useful relevant observational information so that anyone can use them.


Fits files can be split into different "extensions". These extensions contain different information. Different telescopes and processed data have a different number of extensions and with different data is each one.

Below is an example of how to open the FITS files in python.

You will need to change the path to match. This file can be found in the zip folder from the LEARN page titled: "Lab 2 Files - FITS Lab & Crowded Field"

### Saving the files



```python
# Save a brand new fits file

data_to_save = np.random.random((100, 100))
hdu = fits.PrimaryHDU(data_to_save)
hdul = fits.HDUList([hdu])
hdul[0].header['FILTER'] = 'White'
hdul.writeto('new_file.fits', overwrite=True)
hdul.close()

# Rewriting a FITS file
hdul = fits.open(folder + filename, mode='update')
data = hdul[1].data
modified_data = data + 1000
hdul[1].data = modified_data
hdul[0].header['COMMENT'] = 'Modified with Python'
hdul.flush()

# Saving a new FITS file
hdul = fits.open(folder + filename, mode='update')
data = hdul[1].data
modified_data = data + 1000
hdul[1].data = modified_data
hdul[0].header['COMMENT'] = 'Modified with Python'
hdul.writeto('output.fits', overwrite=True)
hdul.close()
```
